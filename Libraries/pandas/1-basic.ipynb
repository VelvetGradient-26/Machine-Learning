## Dataframes

```python
np.random.seed(42)
randoms = np.random.randn(5,4)
randoms
```

```python
rows = ['a', 'b', 'c', 'd', 'e']
columns = ['w', 'x', 'y', 'z']
```

```python
df = pd.DataFrame(data=randoms, index=rows, columns=columns)
df
```

```python
df[['x', 'y']]
```

```python
# create a new column
df['new'] = df['w'] + df['y']
df
```

```python
# delete a column
df.drop("new", axis=1, inplace=True)
df
```

```python
df.drop('a', inplace=True, axis=0)
df
```

```python
# locate row by named index
df.loc["b"]
```

```python
# locate row by numeric index
df.iloc[1]
```

```python
# locate multiple rows
df.loc[['b', 'c', 'd']]
```

```python
# select specific rows and columns
df.loc[['b', 'c'], ['y', 'z']]
```
