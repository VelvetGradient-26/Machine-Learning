```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
arr = np.random.randn(5,4)
arr
```

```python
rows = ['a', 'b', 'c', 'd', 'e']
columns = ['w', 'x', 'y', 'z']
```

```python
df = pd.DataFrame(data=arr, index=rows, columns=columns)
df
```

## Conditional Filters

```python
bool_df = df > 0
bool_df
```

```python
df[bool_df]
```

```python
# select rows with only values greater than zero in the column 'w'
df[df['w'] > 0]
```
